### EDA and official implicit conversion

Every rating from 1 to 5 is an observed positive interaction. Raw ratings are retained for history and UI display.

```python
from pathlib import Path
import sys
import json
import pandas as pd
import torch

ROOT = Path.cwd().resolve().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))
PROCESSED = ROOT / "data" / "processed"
MODELS = ROOT / "models"
RESULTS = ROOT / "results"
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("device:", DEVICE)
```

```python
from src.data.implicit import load_movielens_ratings
from src.data.implicit import build_temporal_implicit_splits

raw = load_movielens_ratings(ROOT / "data/raw/ml-1m/ratings.dat")
train, validation, test, eligible = build_temporal_implicit_splits(raw, min_interactions=3)
print("raw:", raw.shape)
print("eligible users:", eligible.user_id.nunique())
print("train/validation/test:", len(train), len(validation), len(test))
```

```python
for name, frame in (("train", train), ("validation", validation), ("test", test), ("eligible_interactions", eligible)):
    frame.to_csv(PROCESSED / f"{name}.csv", index=False)
print("saved official implicit splits")
```
